# 준비작업 및 데이터 가져오기

```python



# -------------------------------
# 데이터 처리
# -------------------------------
import numpy as np
from pandas import DataFrame


# -------------------------------
# 시각화
# -------------------------------
import seaborn as sb
from matplotlib import pyplot as plt


# -------------------------------
# 머신러닝 / 평가 지표
# -------------------------------
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, silhouette_samples
from sklearn.cluster import AgglomerativeClustering


# -------------------------------
# 기하 / 계층적 군집
# -------------------------------
from scipy.spatial import ConvexHull
from scipy.cluster.hierarchy import dendrogram, linkage
# -------------------------------
# 덴드로그램 구현 함수
# -------------------------------
from scipy.cluster.hierarchy import dendrogram
```

```python
from hossam import load_data, my_dpi, visualize_silhouette
origin = load_data("iris")
origin.head()
```

# 최적의 군집 수 찾기

### 종속변수 제거 + 데이터 스케일링

```python
# 타겟 컬럼 제거
df = origin.drop("Species", axis=1)

# 표준화
scaler = StandardScaler()
sdf = DataFrame(
    scaler.fit_transform(df),
    columns=df.columns
)

# 결과 확인
sdf.head()
```

# 최적의 군집 수 찾기

### 군집수 바꿔가며 실루엣 스코어 확인

```python
# -------------------------------
# 군집 수 후보
# -------------------------------
klist = list(range(3, 11))
s_items = []


# -------------------------------
# k별 Agglomerative Clustering
# -------------------------------
for k in klist:

    estimator = AgglomerativeClustering(
        n_clusters=k,
        # 노드 간 거리 계산값 도출 여부 (시각화를 위해 필수)
        compute_distances=True,
    )

    # 모델 학습
    estimator.fit(sdf)

    # 군집 결과
    cluster = estimator.labels_

    # 실루엣 시각화
    visualize_silhouette(estimator, sdf)

    # 실루엣 스코어 계산
    s_score = silhouette_score(
        X=sdf,
        labels=cluster
    )

    # 결과 저장
    s_items.append(
        {
            "k": k,
            "score": s_score
        }
    )


# -------------------------------
# 결과 정리
# -------------------------------
score_df = DataFrame(s_items)

score_df.sort_values(
    by="score",
    ascending=False,
    inplace=True
)

score_df
```

# 최종 모형 적합

```python
# -------------------------------
# Agglomerative Clustering 설정
# -------------------------------
estimator = AgglomerativeClustering(
    n_clusters=best_k,
    compute_distances=True,
)


# -------------------------------
# 모델 학습
# -------------------------------
estimator.fit(sdf)


# -------------------------------
# 결과 정리
# -------------------------------
result_df = df.copy()
result_df["cluster"] = estimator.labels_

result_df.head()
```

### 시각화

```python
# -------------------------------
# Dendrogram 데이터 생성
# -------------------------------
from dandrogram_source import dandrogram_source

data = dandrogram_source(estimator)


# -------------------------------
# Figure 설정
# -------------------------------
figsize = (1500 / my_dpi, 800 / my_dpi)
fig, ax = plt.subplots(
    nrows=1,
    ncols=1,
    figsize=figsize,
    dpi=my_dpi
)


# -------------------------------
# Dendrogram 시각화
# -------------------------------
dendrogram(
    data,
    ax=ax,
    p=30,
    truncate_mode="lastp",
    leaf_rotation=0,
    leaf_font_size=8,
    count_sort="ascending",
)


# -------------------------------
# 그래프 옵션 및 출력
# -------------------------------
ax.set_title("Dendrogram", fontsize=12, pad=8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
plt.close()
```
