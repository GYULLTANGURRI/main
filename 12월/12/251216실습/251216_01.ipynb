📝2.결과보고 함수 정의

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from statsmodels.stats.outliers_influence import variance_inflation_factor


def logit_report(fit, data, threshold=0.5):
    """
    Logit summary(tbl)와 원 데이터프레임(data)을 받아
    1) 성능지표 DataFrame (cdf)
    2) 변수 요약 DataFrame (rdf)
    3) 모델 요약 문장 (result_report)
    4) 모형 설명 문장 (model_report)
    5) 변수별 설명 문장 리스트 (variable_reports)
    를 반환한다.
    """

    # =================================================
    # 1. 성능 평가 지표
    # =================================================
    yname = fit.model.endog_names
    y_true = data[yname].values

    y_prob = fit.predict(fit.model.exog)
    y_pred = (y_prob >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

    acc = accuracy_score(y_true, y_pred)
    pre = precision_score(y_true, y_pred)
    tpr = recall_score(y_true, y_pred)
    fpr = fp / (fp + tn)
    tnr = 1 - fpr
    f1 = f1_score(y_true, y_pred)
    auc = roc_auc_score(y_true, y_prob)

    cdf = pd.DataFrame({
        "설명력(Pseudo R²)": [fit.prsquared],
        "정확도(Accuracy)": [acc],
        "정밀도(Precision)": [pre],
        "재현율(Recall, TPR)": [tpr],
        "위양성률(FPR)": [fpr],
        "특이성(TNR)": [tnr],
        "ROC AUC": [auc],
        "F1-score": [f1]
    })

    # =================================================
    # 2. 회귀계수 요약 (OR 중심)
    # =================================================
    tbl = fit.summary()
    xnames = [n for n in fit.model.exog_names if n != "const"]
    X = data[xnames]

    # VIF 계산
    vif_dict = {
        col: variance_inflation_factor(X.values, i)
        for i, col in enumerate(X.columns)
    }

    variables = []

    for v in tbl.tables[1].data:
        name = v[0].strip()
        if name not in xnames:
            continue

        beta = float(v[1])
        se = float(v[2])
        p = float(v[4])

        or_val = np.exp(beta)
        ci_low = np.exp(beta - 1.96 * se)
        ci_high = np.exp(beta + 1.96 * se)

        stars = (
            "***" if p < 0.001 else
            "**" if p < 0.01 else
            "*" if p < 0.05 else
            ""
        )

        variables.append({
            "종속변수": yname,
            "독립변수": name,
            "B(β)": beta,
            "표준오차": se,
            "z": float(v[3]),
            "p-value": p,
            "OR": or_val,
            "CI_lower": ci_low,
            "CI_upper": ci_high,
            "VIF": vif_dict.get(name, np.nan),
            "유의성": stars
        })

    rdf = pd.DataFrame(variables)

    # =================================================
    # 3. 모델 요약 문장
    # =================================================
    result_report = (
        f"Pseudo R²(McFadden) = {fit.prsquared:.3f}, "
        f"LLR χ²({int(fit.df_model)}) = {fit.llr:.3f}, "
        f"p-value = {fit.llr_pvalue:.4f}, "
        f"Accuracy = {acc:.3f}, "
        f"AUC = {auc:.3f}"
    )

    tpl = (
        "%s에 대하여 %s로 예측하는 로지스틱 회귀분석을 실시한 결과, "
        "모형은 통계적으로 %s(χ²(%s) = %.3f, p %s 0.05)하였다."
    )

    model_report = tpl % (
        yname,
        ", ".join(xnames),
        "유의" if fit.llr_pvalue <= 0.05 else "유의하지 않음",
        int(fit.df_model),
        fit.llr,
        "<=" if fit.llr_pvalue <= 0.05 else ">"
    )

    # =================================================
    # 4. 변수별 해석 문장
    # =================================================
    variable_reports = []

    s = (
        "%s의 오즈비는 %.3f(p %s 0.05)로, "
        "%s 발생 odds에 %s 영향을 미치는 것으로 나타났다."
    )

    for _, row in rdf.iterrows():
        variable_reports.append(
            s % (
                row["독립변수"],
                row["OR"],
                "<=" if row["p-value"] < 0.05 else ">",
                row["종속변수"],
                "유의한" if row["p-value"] < 0.05 else "유의하지 않은"
            )
        )

    return cdf, rdf, result_report, model_report, variable_reports
```

📝분석결과 보고

```python
cdf, rdf, r, m ,v = logit_report(fit,origin)
display(cdf)
display(rdf)
display(r)
display(m)
display(v)
```

로지스틱에서 VIF는 참고용일 뿐, VIF가 10을 넘는다고 해서 제거할 필요는 없음<br>
VIF(분산팽창계수) 자체가 선형성일때 사용되는 값이기 때문이다<br>
즉, 로지스틱에서 VIF 가 10을 넘는다는 것은 "이 변수가다른 독립변수들과 강하게 선형적으로 얽혀 있다"는 뜻이지,<br>
"이 변수가 틀렸다/쓸 수 없다"는 의미는 아니다.

2.독립변수 로그 변환

📝1.로그 변환 후 다시 분석

```python
# ===============================
# 1. 데이터 복사
# ===============================
df = origin.copy()


# ===============================
# 2. 종속 / 독립 변수 분리
# ===============================
yname = "Outcome"
y = df[yname]
X = df.drop(columns=[yname])


# ===============================
# 3. 로그변환 대상 변수 (연속형 예시)
# ===============================
log_vars = X.columns


# ===============================
# 4. 로그 변환 (0 이하 값 방지)
# ===============================
for v in log_vars:
    df[f"log_{v}"] = np.log1p(df[v])


# ===============================
# 5. 원 변수 제거
# ===============================
df.drop(columns=log_vars, inplace=True)


# ===============================
# 6. 최종 입력 데이터 구성
# ===============================
X_log = df.drop(columns=[yname])
X_input = sm.add_constant(X_log)


# ===============================
# 7. 로지스틱 회귀 적합
# ===============================
model = sm.Logit(y, X_input)
fit = model.fit()


print(fit.summary())
print(f"상수항은 모든 독립변수 (x)가 0일 때의 y값"
      f"포도당, 체질량지수, 나이 등의 계수가 양수 그리고 값도 꽤 큼"
      f"실제 데이터에서 사람들의 혈당,BMI,나이는 0\n로그를 취해도 일정한 양수 값을 가짐\n하지만 결과값(y)은 특정 범위에 있어야 함.\n독립변수들에 의해 계산된 값이 너무 크기 때문에, 실제 결과값 수준으로 맞추기 위해 상수항을 깎아내림./n그리하여 매우 큰 음수 -26이 됨")

```
