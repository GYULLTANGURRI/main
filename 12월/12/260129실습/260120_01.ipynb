```python
# -------------------------------
# 사용자 정의 모듈
# -------------------------------
from hossam import load_data, my_dpi, visualize_silhouette


# -------------------------------
# 데이터 처리
# -------------------------------
import numpy as np
from pandas import DataFrame


# -------------------------------
# 시각화
# -------------------------------
import seaborn as sb
from matplotlib import pyplot as plt


# -------------------------------
# 머신러닝 / 평가 지표
# -------------------------------
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, silhouette_samples
from sklearn.cluster import AgglomerativeClustering


# -------------------------------
# 기하 / 계층적 군집
# -------------------------------
from scipy.spatial import ConvexHull
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
origin = load_data("iris")
origin.head()
```

```python
df = origin.copy()
# 1. 전체 통계 보기 (describe().T)
print(df.describe().T)

# 2. 품종별 꽃받침 너비(Sepal Width) 평균
result = df.groupby('Species')['Sepal.Width'].mean()
print("\n품종별 꽃받침 너비 평균:")
print(result)

# 3. 품종별 통계
print("\n품종별 꽃받침 너비 상세 통계:")

print(df.groupby('Species')['Sepal.Width'].describe().T)
```
