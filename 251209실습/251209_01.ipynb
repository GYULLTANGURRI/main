📝 [2] 데이터 분포 시각화<br>
(1) 관수빈도별 햇빛 노출 빈도에 따른 식물 높이 비교<br>

```python
field = 'height'
group_field = 'water'
hue_field = 'sun'
# 시각화 할 필드만 원본 데이터에서 추출
data = origin[[field, group_field, hue_field]]
group_values = data[group_field].unique()
group_count = len(group_values)
hue_values = data[hue_field].unique()
# 1) 그래프 초기화
rows = group_count # 그래프 행 수
cols = 1 # 그래프 열 수
width_px = 1280 * cols # 그래프 가로 크기
height_px = 640 * rows # 그래프 세로 크기
figsize = (width_px / my_dpi, height_px / my_dpi)
# ax 객체가 행,열 수에 따라서 리스트가 된다.
fig, ax = plt.subplots(rows, cols, figsize=figsize, dpi=my_dpi)
# 2) 그래프 그리기
for i, v in enumerate(group_values):
    temp = data[data[group_field] == v]
    sb.kdeplot(data=temp, x=field, hue=hue_field, ax=ax[i])
    ax[i].grid(True, alpha=0.3)
    ax[i].set_title('{0} = {1}'.format(group_field, v), fontsize=10,
        pad=5)
    for h in hue_values:
        temp_sub = temp[temp[hue_field] == h]
# 신뢰구간
    max = temp_sub[field].max() # 최대값
    clevel = 0.95 # 신뢰수준
    dof = len(temp_sub[field]) - 1 # 자유도
    sp_mean = temp_sub[field].mean() # 표본평균
    sp_std = temp_sub[field].std(ddof=1) # 표본표준편차
    sp_std_error = sp_std / sqrt(len(temp_sub[field])) # 표본표준오차
    cmin, cmax = t.interval(clevel, dof, loc=sp_mean,
    scale=sp_std_error)
    ax[i].axvline(cmin, linestyle=':', linewidth=0.5)
    ax[i].axvline(cmax, linestyle=':', linewidth=0.5)
    # 평균 표시
    ymin, ymax = ax[i].get_ylim()
    ax[i].fill_between([cmin, cmax], 0, ymax, alpha=0.1)
    ax[i].axvline(sp_mean, linestyle='--', linewidth=1)
    # 텍스트 그리기
    ax[i].text(x=(cmax-cmin)/2+cmin,
    y=ymax,
    s="[%s] %0.1f ~ %0.1f" % (h, cmin, cmax),
            horizontalalignment="center",
            verticalalignment="bottom",
            fontdict={"size": 6, "color": "red"})
# 4) 출력
plt.tight_layout() # 여백 제거
plt.show() # 그래프 화면 출력
plt.close() # 그래프 작업 종료
```

📝 [3] 분산분석의 가정 확인

```python
data = origin.copy()
xname = ['water', 'sun']
yname = 'height'
u1 = data[xname[0]].unique()
u2 = data[xname[1]].unique()
equal_var_fields = []
normal_dist = True
report = []
for i in u1:
    for j in u2:
        filtered_data = data[(data[xname[0]] == i) & (data[xname[1]]
        == j)][yname]
        equal_var_fields.append(filtered_data)
        s, p = normaltest(filtered_data)
        normalize = p > 0.05
        report.append({
        "field": "{0}, {1}".format(i, j),
        "statistic": s,
        "p-value": p,
        "result": normalize
        })
        normal_dist = normal_dist and normalize
if normal_dist:
    n = "Bartlett"
    s, p = bartlett(*equal_var_fields)
else:
    n = "Levene"
    s, p = levene(*equal_var_fields)
equal_var = p > 0.05
report.append({
    "field": n,
    "statistic": s,
    "p-value": p,
    "result": p > 0.05
})
report_df = DataFrame(report).set_index('field')
report_df
```

📝 [4] 이원분산분석<br>
1) 분산분석 결과 생성<br>

```python
xname = ['water', 'sun']
yname = 'height'
pg_anova = anova(data=origin, dv=yname, between=xname)
pg_anova
```

2) 결과 문장 생성

```python
for idx in pg_anova.index:
    row = pg_anova.iloc[idx]
    source = row['Source']
    p_val = row['p-unc']
    if p_val == 0.05:
        print(f"→ {source} 효과는 유의함. (대립가설 채택: p ≤ 0.05)")
else:
    print(f"→ {source} 효과는 유의하지 않음. (귀무가설 채택: p > 0.05)")
```

ANOVA는 데이터를 “설명되는 부분 + 설명되지 않는 부분”으로 나눈다<br>
총제곱합(SS Total) = 요인 A(water)로 설명되는 부분 + 요인 B(sun)으로 설명되는 부<br>
분 + 상호작용(water × sun)으로 설명되는 부분 + 남은 부분(Residual)<br>

Residual = “모델이 설명할 수 없는 개별 오차”<br>
동일 조건인데도 식물 키(height)가 조금씩 다른 자연 변동<br>
측정 오차<br>
눈에 보이지 않는 다른 요인들(토양의 미세 차이, 개체 특성 등)<br>
모델에서는 고려하지 않은 모든 외부 영향<br>
Residual은 DF(자유도)와 함께 모델의 품질을 보여줌<br>
Residual[SS] = 107.264<br>
Residual[DF] = 114<br>
Residual[MS] = 0.941<br>
이 말은,<br>
각 조건에서 개체 간 차이는 평균적으로 약 0.94 정도의 변동을 가진다는 뜻<br>
모델이 설명한 변동이 훨씬 크므로 주효과(water, sun)의 F값이 매우 큼<br>
즉, Residual이 작을수록 모델 설명력이 높다는 의미임<br>
Residual은 ANOVA 모델이 설명하지 못한 오차(개체 간 자연 변동 + 측정 오차 등)를 의<br>
미하며, F-통계량을 계산하는 기준이 되는 값이다.<br>

3) 사후검정

```python
xname = ['water', 'sun']
yname = 'height'
data = origin[xname + [yname]]

# 요인별 수준 수 계산
levels = {}
# 예: {'water': 2, 'sun': 3}
for factor in xname:
    level_count = data[factor].nunique()
    levels[factor] = level_count

# ANOVA p-value 사전 만들기
pvals = {}
# 예: {'water': 1.30e-20, 'sun': 2.01e-17, 'water * sun': 0.105}
for index, row in pg_anova.iterrows():
    source_name = row['Source']
    p_value = row['p-unc']
    pvals[source_name] = p_value

# 주효과 처리
for factor in xname:
    print(f"\n[{factor}]")

    # 수준 수 2인 경우
    if levels[factor] == 2:
        print(f"- {levels[factor]}수준 요인이므로 사후검정이 필요 없습니다.")
        continue

    # p-value 확인
    p_val = pvals.get(factor, None)
    if p_val is None:
        print("- ANOVA 결과에서 해당 요인을 찾을 수 없습니다.")
        continue

    if p_val > 0.05:
        print(f"- 주효과 유의하지 않음 (p={p_val:.3g}) → 사후검정 수행 안 함")
        continue

    # 유의한 경우
    print(f"- 주효과 유의 (p={p_val:.3g}) → 사후검정 수행")

    if equal_var:
        print(" · Tukey HSD 사용 (등분산 충족)")
        hoc_df = pairwise_tukey(data=data, dv=yname, between=factor)
    else:
        print(" · Games–Howell 사용 (등분산 불충족)")
        hoc_df = pairwise_gameshowell(data=data, dv=yname, between=factor)

    display(hoc_df)

# 상호작용 자동 처리
interaction_names = [src for src in pvals.keys() if '*' in src]

for inter in interaction_names:
    p_val = pvals[inter]
    print(f"\n[{inter}] 상호작용 효과")

    if p_val <= 0.05:
        print(f"- 상호작용 유의 (p={p_val:.3g}) → 단순주효과 분석 필요")
    else:
        print(f"- 상호작용 유의하지 않음 (p={p_val:.3g}) → 단순주효과 분석 필요없음")
```

📝 1. 분석 목적<br>
본 분석은 식물의 높이(height)에 대해 관수 빈도(water)와 햇빛 노출(sun)이 어떤 영향을 미<br>
치는지 평가하기 위해 이원분산분석(two-way ANOVA)을 수행하였다.<br>

📝 2. 분산분석 주요 결과

2-1. 요인별 유의성 판단<br>
water: F = 130.53, p = 1.30e-20, ηp² = 0.534 → 유의함(대립가설 채택)<br>
sun: F = 54.89, p = 2.01e-17, ηp² = 0.491 → 유의함(대립가설 채택)<br>
water × sun: F = 2.30, p = 0.105, ηp² = 0.039 → 유의하지 않음(귀무가설 채<br>
택)<br>

물주기와 햇빛은 식물 높이에 모두 유의한 영향을 주지만, 두 요인이 결합해 특별한 상호작용 효
과를 만들어내지는 않는다.

📝 3. 기술통계 및 집단별 평균 비교<br>
water 수준별 평균<br>
daily: 6.63<br>
weekly: 4.57
매일 물 준 그룹이 크게 성장함.<br>
sun 수준별 평균<br>
low: 4.44<br>
med: 5.61<br>
high: 6.73<br>
햇빛이 많을수록 뚜렷한 증가.<br>
두 요인의 증가 방향은 거의 평행한 형태로 나타나 상호작용이 약하다는 점을 뒷받침한다.<br>

📝 4. 사후검정(Post-hoc) 결과<br>
water: 2수준 요인이므로 사후검정 불필요.<br>
sun: 3수준 + 주효과 유의 → Tukey HSD 수행<br>
결과:<br>
high > low (차이 2.27, p < .001)<br>
high > med (차이 1.12, p < .01)<br>
med > low (차이 1.15, p < .01)<br>
상호작용: p=0.105 → 단순주효과 분석 불필요<br>
햇빛의 세 조건(low, med, high)은 서로 모두 유의하게 다른 평균을 가진다.<br>

📝 5. 최종 결론 요약<br>
관수(water)는 식물의 성장에 강력한 영향을 줌(매일 물을 줄수록 더 큼).<br>
햇빛(sun)은 명확한 선형적 효과를 가지며, high > med > low 순으로 성장 증가.<br>
두 요인이 결합해 상승효과를 만드는 상호작용은 없음.<br>
사후검정 결과, 햇빛 조건 세 그룹은 모두 서로 유의하게 차이가 있음.<br>
식물의 높이는 관수 빈도와 햇빛 수준의 영향을 크게 받으며, 두 요인은 독립적으로 작용하여 성<br>
장 패턴을 결정한다.<br>
